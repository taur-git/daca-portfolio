# Roll A — Data Loading

```python
import os
from dotenv import load_dotenv

load_dotenv()

SUPABASE_URL = os.getenv("SUPABASE_URL")
SUPABASE_KEY = os.getenv("SUPABASE_KEY")

print("URL exists:", SUPABASE_URL is not None)
print("KEY exists:", SUPABASE_KEY is not None)
```

```python
import pandas as pd
from supabase import create_client
```

```python
supabase = create_client(SUPABASE_URL, SUPABASE_KEY)

print("Supabase connected")
```

```python

# Kqigi andmete laadimine sales tabelist
all_sales = []
offset = 0
while True:
    response = supabase.table('sales').select('*').range(offset, offset + 999).execute()
    if not response.data:
        break
    all_sales.extend(response.data)
    offset += 1000

df_sales = pd.DataFrame(all_sales)
print("Sales shape:", df_sales.shape)
df_sales.head()
```

```python
# Kqigi andmete laadimine customers tabelist
all_customers = []
offset = 0
while True:
    response = supabase.table('customers').select('*').range(offset, offset + 999).execute()
    if not response.data:
        break
    all_customers.extend(response.data)
    offset += 1000

df_customers = pd.DataFrame(all_customers)
print("Customers shape:", df_customers.shape)
df_customers.head()
```

```python
df_merge = pd.merge(
    df_sales,
    df_customers,
    on="customer_id",
    how="left"
)

print("Merged shape:", df_merge.shape)

df_merge.head()
```

```python
print("Merged DataFrame shape:")
print(df_merge.shape)

print("\nData types:")
print(df_merge.dtypes)

print("\nFirst 5 rows:")
display(df_merge.head())
```

# ROLL C - FM kliendisegmenteerimine

```python
today = pd.to_datetime('2026-05-12')
```

```python
recency = df_merge.groupby('customer_id')['sale_date'].max().reset_index()
recency.columns = ['customer_id', 'last_purchase']
recency['recency_days'] = (today - recency['last_purchase']).dt.days
print(recency.head())
```

```python
frequency = df_merge.groupby('customer_id')['sale_id'].count().reset_index()
frequency.columns = ['customer_id', 'frequency']
print(frequency.head())
```

```python
monetary = df_merge.groupby('customer_id')['total_price'].sum().reset_index()
monetary.columns = ['customer_id', 'monetary']
print(monetary.head())
```

```python
# Liidan R, F, M kokku
rfm = pd.merge(recency, frequency, on='customer_id')
rfm = pd.merge(rfm, monetary, on='customer_id')

# Skoorid 1-5
rfm['R_score'] = pd.qcut(rfm['recency_days'], q=5, labels=[5,4,3,2,1])
rfm['F_score'] = pd.qcut(rfm['frequency'].rank(method='first'), q=5, labels=[1,2,3,4,5])
rfm['M_score'] = pd.qcut(rfm['monetary'].rank(method='first'), q=5, labels=[1,2,3,4,5])

print(rfm.head())
```

```python
# RFM koondskoor
rfm['RFM_score'] = rfm['R_score'].astype(int) + rfm['F_score'].astype(int) + rfm['M_score'].astype(int)

# Segmendid
def segment(score):
    if score >= 13:
        return 'VIP Champions'
    elif score >= 10:
        return 'Loyal'
    elif score >= 7:
        return 'Potential'
    elif score >= 4:
        return 'At Risk'
    else:
        return 'Lost'

rfm['segment'] = rfm['RFM_score'].apply(segment)

# Kokkuvõte
print(rfm['segment'].value_counts())
```

```python
# Kas kõik kliendid said segmendi
print(f"Kokku kliente: {rfm.shape[0]}")
print(f"Segmendita kliente: {rfm['segment'].isnull().sum()}")

# Segmentide jaotus
print("\nSegmentide jaotus:")
print(rfm['segment'].value_counts())

# Protsentides
print("\nProtsentides:")
print(rfm['segment'].value_counts(normalize=True).mul(100).round(1).astype(str) + '%')
```

```python
import plotly.express as px
```

```python
# Segmentide klientide arv
segment_data = rfm['segment'].value_counts().reset_index()
segment_data.columns = ['segment', 'klientide_arv']

fig = px.bar(
    segment_data,
    x='segment',
    y='klientide_arv',
    title='Diagramm 1. UrbanStyle segmentide klientide arv',
    labels={'segment': 'Kliendisegment', 'klientide_arv': 'Klientide arv (tk)'},
    color_discrete_sequence=['#009B8D'],
    text_auto=True
)

fig.update_layout(
    plot_bgcolor='#F5F5F5'
)

fig.update_traces(
    textfont_color='#1A1A2E'
)
fig.update_layout(
    height=650,
    width=1200
)
fig.show()
```
